```python
""" Find who earns the most money in each of the company's departments.
A higher earner in a department is an employee who has a salary on the top three unique salaries
for that department

"""
emp_data = [
(1,"joe", 85000, 1),
(2,"Henry", 80000, 2),
(3,"Sam", 60000, 2),
(4,"Max", 90000, 1),
(5,"Janet", 69000, 1),
(6,"Randy", 85000, 1),
(7,"Will", 70000, 1)
]
from pyspark.sql import SparkSession
from pyspark.sql.window import Window
from pyspark.sql.functions import dense_rank,col
spark = SparkSession.builder.appName("high_salary").getOrCreate()
schema = ["emp_id","name","salary","dept_id"]
df = spark.createDataFrame(emp_data,schema=schema)
windowspec = Window.partitionBY("dept_id").orderbBydesc("salary")
```
